#K-means

image.png

K-Means is an unsupervised clustering algorithm that groups data into K clusters by minimizing how far data points are from their cluster’s center.

#Steps

1- Step 1: Choose K

2- Step 2: Initialize centroids

3- Step 3: Assignment step

4- K-Means minimizes Within-Cluster Sum of Squares (WCSS)

5- if (WCSS) large , Repeat the task agin with the updated centroids(Average)

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_blobs
from sklearn.metrics import silhouette_score

import pandas as pd

```

```python
df=pd.read_csv("/content/make_blobs_dataset.csv")
```

image.png

```python
df.head(2)
```

```python
X=df
```

Feature Scaling

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

#Call the Model

```python
kmeans = KMeans(
    n_clusters=2,        # K
    init='k-means++',    # smart initialization
    max_iter=300,        # max iterations per run
    random_state=42
)

kmeans.fit(X_scaled)
```

#K-Means++
is a smart initialization method for K-Means. Instead of selecting the initial centroids randomly, it chooses them far apart from each other. This usually leads to faster convergence and better clustering results.

#what is the cluster result ?


```python
labels = kmeans.labels_                 # cluster assignment for each point
centroids = kmeans.cluster_centers_     # final centroids
inertia = kmeans.inertia_               # WCSS
```

```python
inertia
```

```python
centroids
```

```python
labels
```

```python
X_scaled
```

# Can we See the grps ?

```python
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels)
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    marker='X',
    s=200,
    c='red'
)
plt.title("K-Means Clustering")
plt.show()
```

#Evaluate clustering — Silhouette Score

```python
score = silhouette_score(X_scaled, labels)
print("Silhouette Score:", score)
```

### this is risk model because you should choose the right number of grps  (K)

###So if We want To choose the right K use : Elbow Method

#what is Elbow Method ?

(WCSS) VS (K)

#Elbow Method


image.png

image.png

image.png

```python
wcss = []

# k =[ 1 , 2 , .....10]
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss)
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
df=pd.read_csv("/content/iris.csv")
```

```python
df.head(2)
```

```python
X=df.drop('target',axis=1)
y=df['target']
X
```

```python
y
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss)
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method (Iris Dataset)")
plt.show()
```

```python
kmeans = KMeans(
    n_clusters=4,
    init='k-means++',
    max_iter=300,
    random_state=42
)

kmeans.fit(X_scaled)
```

```python
labels = kmeans.labels_               # cluster assignment
centroids = kmeans.cluster_centers_   # cluster centers
inertia = kmeans.inertia_             # WCSS
```

```python
sil_score = silhouette_score(X_scaled, labels)
print("Silhouette Score:", sil_score)
```

```python
plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=labels
)

plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    marker='X',
    s=200
)

plt.xlabel(df.columns[0])
plt.ylabel(df.columns[1])
plt.title("K-Means on Iris Dataset")
plt.show()
```
