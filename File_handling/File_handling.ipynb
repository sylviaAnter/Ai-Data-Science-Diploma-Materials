create a file of numbers then get the avr of these numbers

```python
f=open("numbers.txt", "w")
f.write("1\n2\n3\n4\n5")
f.close()

#-------------------------

f=open("numbers.txt", "r")
total=0
for line in f:
    total += int(line)

f.seek(0)  # Move the file pointer back to the beginning of the file
count = len(f.readlines())
avr = total / count 
print("Average of numbers in file is: ", avr)
f.close()
```

-----------------------

with statement : open the file apply the operations then close automatically 

```python
f=open("numbers.txt", "r")
print(f.read())
f.close()

#---------------------------------------

with open("numbers.txt", "r") as f:
    print(f.read())
```

```python
with open("new_file.txt", "a") as f2:
    f2.write("aaaaa\n")
    f2.write("aaaaa\n")
    f2.write("aaaaa\n")
    f2.write("aaaaa\n")
```

w+ (Write and Read)

Creates the file if it doesn't exist.

If the file exists, everything is erased immediately.

Cursor starts at the beginning.

```python
file = open("students2.txt", "w+")

file.write("Sara\n")
file.write("mona\n")
file.write("soha\n")

file.seek(0)
print(file.read())

file.seek(0)
file.write("yyyy\n")
file.write("yyyy\n")
file.write("yyyy\n")
file.seek(0)
print(file.read())
file.close()
```

a+ (Append and Read)

Creates the file if it doesn't exist.

Keeps all existing content.

Every write is added to the end of the file.

```python
file = open("students2.txt", "a+")

file.write("Sandy")

file.seek(0)
print(file.read())

file.close()
```
