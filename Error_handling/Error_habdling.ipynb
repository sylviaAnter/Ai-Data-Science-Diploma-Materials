```python
number1=int(input("Enter first number: "))
number2=int(input("Enter second number: "))

try:
    print(number1/number2)
except:
    print("Error.")

print('end the program')
print(2+2)
```

```python
number1=22
number2=0

try:
    print(number1/number2)
except ZeroDivisionError:
    print("Error.")

print('end the program')
print(2+2)
```

```python
number1=int(input("Enter first number: "))
number2=int(input("Enter second number: "))

try:
    print(number1/number2)
except Exception as e:
    print(f"Error is  {e}")

print('end the program')
print(2+2)
```

```python
try:
    number1 = int(input("Enter first number: "))
    number2 = int(input("Enter second number: "))
except Exception as e:
    print(f"Error is : {e}")

except:
    print("unexpected error")    

print(number1 - number2)
```

```python
number1=int(input("Enter first number: "))
number2=int(input("Enter second number: "))

try:
    print(number1/number2)
except Exception as e:
    print(f"Error is  {e}")

else:
    print("No error")

print('end the program')
print(2+2)
```

```python
number1=int(input("Enter first number: "))
number2=int(input("Enter second number: "))

try:
    print(number1/number2)
except Exception as e:
    print(f"Error is  {e}")

else:
    print("No error")

finally:
    print("This will always execute")

print('end the program')
print(2+2)
```
